O método describe é uma ferramenta interessante para obter informações estatísticas do dataframe que estamos analisando. Em particular, quando combinado com os métodos loc, iloc ou query, esse método torna-se bastante poderoso na análise de dados massivos. Considerando o dataframe mostrado anteriormente, use a combinação de métodos loc/iloc/query + describe() para responder a seguinte pergunta: qual a média (mean) e desvio padrão (std) de votos recebidos pelo partido PSK no primeiro turno das três eleições mostradas?

```python
# Filtrando o dataframe para o partido PSK no primeiro turno
psk_first_turn = df.query("Partido == 'PSK' & Turno == 1")

# Calculando a média e o desvio padrão dos votos recebidos pelo partido PSK no primeiro turno
statistics = psk_first_turn.describe().loc[['mean', 'std'], 'Votos']

print("Média de votos recebidos pelo partido PSK no primeiro turno:")
print(statistics['mean'])

print("\nDesvio padrão de votos recebidos pelo partido PSK no primeiro turno:")
print(statistics['std'])
```
