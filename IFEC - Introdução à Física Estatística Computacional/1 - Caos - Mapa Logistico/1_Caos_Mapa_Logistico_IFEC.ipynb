# Uma breve incursão pelo Caos

## Questão 1
Faça gráficos de yn como função de n para r = 0, 5. Use y0 = 0, 5 e n entre 1 e 50.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Definindo a função para a equação logística
def logistica(r, y0, n_max):
    y_values = [y0]
    for n in range(1, n_max):
        y_next = r * y_values[-1] * (1 - y_values[-1])
        y_values.append(y_next)
    return y_values

# Parâmetros da questão
r = 0.5
y0 = 0.5
n_max = 50

# Gerar os valores de y usando a equação logística
y_values = logistica(r, y0, n_max)

# Plotar o gráfico de y_n como função de n
plt.figure(figsize=(10, 6))
plt.plot(range(n_max), y_values, linestyle='-', color='b')
plt.title(f'Dinâmica Populacional para r = {r}, y0 = {y0}')
plt.xlabel('n (geração)')
plt.ylabel('y_n (população)')
plt.grid(True)
plt.show()
```

## Questão 2
Repita o procedimento acima para os seguintes valores de r: 2.5; 3.1; 3.5 e 3.7. Comente sobre cada resultado. Em particular comente sobre a existência de padrões.

```python
valores_r = [2.5, 3.1, 3.5, 3.7]
y0 = 0.5
n_max = 50

# Gerar gráficos para cada valor de r
plt.figure(figsize=(12, 8))

for i, r in enumerate(valores_r):
    y_values = logistica(r, y0, n_max)
    plt.subplot(2, 2, i+1)
    plt.plot(range(n_max), y_values, linestyle='-', color='b')
    plt.title(f'r = {r}')
    plt.xlabel('n (geração)')
    plt.ylabel('y_n (população)')
    plt.grid(True)

plt.tight_layout()
plt.show()
```

### 𝑟 = 2.5:

O comportamento para 𝑟 = 2.5 é estável. Observa-se que a população converge para um valor fixo após algumas gerações. Esse valor é o ponto de equilíbrio da equação logística, que ocorre em situações onde o crescimento da população é controlado pela capacidade de suporte do ambiente. Não há variações grandes, indicando um padrão previsível.

### 𝑟 = 3.1:

Para 𝑟 = 3.1, começamos a observar um comportamento periódico. Após um número inicial de oscilações, o sistema entra em um ciclo estável de valores periódicos, oscilando entre dois valores. Esse fenômeno é conhecido como "duplicação de período", o que é um indício de que o sistema está se aproximando de um comportamento caótico, mas ainda com algum grau de previsibilidade.

### 𝑟 = 3.5:

Com 𝑟 = 3.5, o comportamento se torna mais complexo e as oscilações periódicas aumentam. O sistema não se estabiliza facilmente, e vemos que a população alterna entre vários valores, entrando em uma espécie de ciclo caótico. A periodicidade aqui começa a desaparecer, com padrões mais imprevisíveis surgindo. Esse é um exemplo clássico de como o caos se inicia em sistemas dinâmicos.

### 𝑟 = 3.7:

Para 𝑟 = 3.7, o comportamento é caótico. As oscilações são altamente imprevisíveis, e pequenas diferenças nas gerações anteriores causam grandes mudanças na evolução subsequente. Não há padrões fixos ou periódicos visíveis, o que caracteriza o caos. Esse é um exemplo claro de como sistemas determinísticos podem apresentar comportamentos complexos e caóticos dependendo das condições.

### Conclusão:

Conforme aumentamos o valor de 𝑟, o sistema passa de um estado estável para um comportamento periódico, até eventualmente exibir caos total. O ponto de transição entre a estabilidade e o caos é caracterizado pela duplicação de períodos, onde o sistema começa a oscilar entre dois valores, depois quatro, oito, e assim por diante, até que a periodicidade se perde completamente. Isso exemplifica a natureza caótica de certos sistemas dinâmicos.

## Questão 3
Refaça os gráficos anteriores considerando três condições iniciais consideravelmente diferentes, ou seja, trace, em um mesmo gráfico, as curvas para y0 = 0.25, y0 = 0.5 e y0 = 0.75. Comente.

𝑟, o sistema é estável e tende a um valor fixo, não sendo muito sensível às condições iniciais. Pequenas mudanças em 𝑦0 não afetam o comportamento final da população, que sempre converge para o mesmo valor.

### 𝑟 = 3.1:

Para 𝑟 = 3.1, há um comportamento periódico e as três condições iniciais 𝑦0 eventualmente entram em sincronia após algumas iterações, convergindo para o mesmo ciclo de dois valores. Isso mostra que, mesmo com diferenças nas condições iniciais, o sistema tende a entrar em um ciclo estável, o que ainda caracteriza um comportamento previsível.

### 𝑟 = 3.5:

Com 𝑟 = 3.5, vemos que as condições iniciais 𝑦0 influenciam o comportamento de maneira mais significativa. Embora haja um comportamento cíclico, as curvas para diferentes 𝑦0 não convergem para um ciclo comum rapidamente. Isso indica que o sistema está se aproximando de um estado caótico, e a influência das condições iniciais começa a se tornar mais pronunciada.

### 𝑟 = 3.7:

Para 𝑟 = 3.7, o sistema é claramente caótico. As três condições iniciais produzem trajetórias completamente diferentes ao longo do tempo, mesmo que comecem próximas. Pequenas variações em 𝑦0 resultam em comportamentos completamente divergentes, ilustrando a característica fundamental do caos: sensibilidade extrema às condições iniciais. Aqui, o sistema não exibe mais padrões periódicos, e o comportamento é imprevisível.

### Conclusão:
* Para valores baixos de 𝑟, o sistema é estável e converge para o mesmo ponto, independentemente de 𝑦0.
* À medida que 𝑟 aumenta, o sistema começa a mostrar comportamento periódico, mas ainda assim as condições iniciais têm pouco impacto no comportamento final.
* Para valores mais altos de 𝑟, o sistema se torna caótico e extremamente sensível às condições iniciais, o que leva a trajetórias completamente diferentes dependendo de 𝑦0.

## Questão 5
Considere 10000 valores de r igualmente espaçados entre 10^−5
e 4. Para cada um desses valores de r, vamos iterar a equação logística 1000 vezes, i.e., n_max = 1000. Em um gráfico de yn como função de r, vamos plotar para cada valor de r os 100 últimos passos da evolução. Teremos 100 pontos no eixo y do gráfico para cada valor de r, que estará no eixo x. A figura gerada é conhecida como diagrama de bifurcação. Essa figura traz inúmeras informações importantes. Tire um tempo para pensar nela e analisar o que encontrou.

```python
# Parâmetros
r_values = np.linspace(1e-5, 4, 10000)  # 10000 valores de r entre 10^-5 e 4
y0 = 0.5  # Condição inicial
n_max = 1000  # Iterações
n_last = 100  # Últimos 100 valores para plotar

# Preparar a figura para o diagrama de bifurcação
plt.figure(figsize=(10, 8))

# Para cada valor de r, iterar a equação logística
for r in r_values:
    y_values = logistica(r, y0, n_max)
    # Plotar os últimos 100 valores de y_n para cada r
    plt.plot([r] * n_last, y_values[-n_last:], ',k', alpha=0.1)  # ',' para um gráfico disperso e rápido

plt.title('Diagrama de Bifurcação da Equação Logística')
plt.xlabel('r')
plt.ylabel('$y_n$ (População)')
plt.grid(False)
plt.show()
```

### O que o diagrama de bifurcação revela:
* **Regiões de estabilidade:** Para valores baixos de 𝑟, o sistema converge para um valor fixo, indicando estabilidade e previsibilidade no comportamento populacional.
* **Duplicações de período:** Conforme 𝑟 aumenta, observamos bifurcações, onde o sistema passa de um valor estável para oscilações periódicas. Este fenômeno é chamado de "duplicação de período", onde o sistema alterna entre dois valores, depois quatro, e assim por diante.
* **Transição para o caos:** Para valores de 𝑟 mais altos (por volta de 𝑟 = 3.5), o sistema entra em um regime caótico, onde pequenos aumentos em 𝑟 resultam em comportamento imprevisível e altamente sensível às condições iniciais.
* **Janelas periódicas:** Dentro da região caótica, há faixas onde o comportamento periódico reaparece. Isso é visível como pequenas ilhas de estabilidade em meio ao caos.

### Conclusão:
O diagrama de bifurcação é uma ferramenta poderosa para visualizar a transição de sistemas dinâmicos de comportamentos estáveis e previsíveis para o caos total. O comportamento caótico pode ser altamente sensível a pequenas variações nos parâmetros, e esse diagrama ilustra claramente essa transição.
